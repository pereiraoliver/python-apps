## Loading the book

```python
import re
```

```python
with open("downloads/miracle_in_the_andes.txt", "r", encoding="utf-8") as file:
    book = file.read()
```

## The most used words (non-articles)

```python
pattern = re.compile(r"[a-zA-Z]+")
findings = re.findall(pattern, book.lower())
findings[:10]
```

```python
dict = {}

for i in findings:
    if i not in dict.keys():
        dict[i] = 1
    else:
        dict[i] = dict[i] + 1
```

```python
d_list = [(values, keys) for keys, values in dict.items()]
d_list = sorted(d_list, reverse=True)
len(d_list)
```

```python
filtered_words = []
for value, key in d_list:
    if key not in english_stopwords:
        filtered_words.append((value, key))

print(filtered_words[:100])
```

## Sentiment Analysis: What is the most positive and the most negative chapter?

### An example

```python
from nltk.sentiment import SentimentIntensityAnalyzer
nltk.download("vader_lexicon")
```

```python
analyzer = SentimentIntensityAnalyzer()
```

```python
# dir(analyzer)
analyzer.polarity_scores("Hey, look how beautiful the trees are. I love them. I really love them")
```

```python
analyzer.polarity_scores(book)
```

### Chapters sentiment analysis

```python
import re
pattern = re.compile(r"Chapter [0-9]+")
# findings = re.findall(pattern, book)
chapters = re.split(pattern, book)
chapters = chapters[1:]
len(chapters)
```

```python
for nr, chapter in enumerate(chapters):
    score = analyzer.polarity_scores(chapter)
    print(nr + 1, score)
```
