## Load the book

```python
import re
with open("downloads/miracle_in_the_andes.txt", "r", encoding="utf-8") as file:
    book = file.read()
book[:2000]
```

## Extract paragraphs where "love" was used

```python
# pattern = re.compile(r"[^ \n]+.*[L|l]ove.*")
pattern = re.compile(r"[^\n]+[L|l]ove[^\n]+")
findings = re.findall(pattern, book)
len(findings)
```

## Extract the chapter titles

```python
# pattern = re.compile(r"Chapter [0-9]+\n+(.*)")
pattern = re.compile(r"([a-zA-Z]+)\n\n")
findings = re.findall(pattern, book)
findings
```

## Function that finds the occurence of any word

```python
def word_count(word):
    pattern = re.compile(rf" ?({word})")
    findings = re.findall(pattern, book.lower())
    if findings:
        print(f"The word \"{word}\" occurs {len(findings)} times.")
    else:
        print(f"The word \"{word}\" is not in the book.")
    return len(findings)
```

## Call the function

```python
word_count("love")
```
